## **📋 Section 1: Architectural Framework & Technical Data Cleaning Overview**

### **🔹 Project Intent & Core Focus**

#### In corporate enterprise environments, raw industrial data lakes are highly volatile and frequently contaminated with processing noise, broken database sequences, missing operational costs, and redundant tracking entries. This production pipeline establishes a multi-layered data purification layout using `Pandas` and `NumPy` to ingest unstable comma-separated value frameworks (.csv) and transition them into a locked, gold-standard analytical asset repository.

#### By designing strict programmatic mitigation layers—such as distribution-safe imputation models, multi-variable transaction deduplication keys, and micro-precision engineered text normalization engines—this codebase ensures an absolute **0% qualification error threshold** across all core business indicators before any downstream modeling or business intelligence tools fetch the array.
---

```python
import pandas as pd
import numpy as np
import os
import warnings

# Suppress runtime slicing and copy warnings for clean console outputs
warnings.filterwarnings('ignore')

# Logic to verify and locate the safe relative file path across environments
paths_to_check = [
    '../data/Dataset_for_Data_Analytics.csv',
    'Dataset_for_Data_Analytics.csv',
    '../Dataset_for_Data_Analytics.csv'
]

working_df = None
for path in paths_to_check:
    if os.path.exists(path):
        working_df = pd.read_csv(path)
        print(f"🌐 Ingestion Success: Data mounted from target branch: '{path}'")
        break

if working_df is None:
    print("⚠️ File path missing! Please verify that 'Dataset_for_Data_Analytics.csv' is inside your 'data' folder.")
else:
    # Create the pristine data processing layer copy right after ingestion
    cleaned_df = working_df.copy()
    print("✅ Success: 'cleaned_df' mirror array initialized for purification steps.")
```

## **🔬 Section 2: Distribution-Safe Strategic Imputation Pipeline (Step 2)**

### **🔹 Core Objective & Technical Logic**

#### Directly dropping missing indices induces statistical selection bias and compromises distribution density. This subsystem maps a mathematical framework using localized central tendency properties to secure missing matrix entries without introducing variance skewing into downstream analytics.

* **Numerical Core Defense:** Overwrites numerical null states using exact feature **Medians** to insulate baseline mathematical layers against extreme record anomalies.

* **Categorical Integrity Guard:** Automatically imputes structural text blanks using high-frequency categorical column **Modes**.

```python
print("🧩 Step 2: Executing distribution-safe strategic imputation arrays...")

# 1. Continuous Numerical Imputation Layer (Utilizing Feature Medians)
numeric_cols = ['Quantity', 'UnitPrice']
for col in numeric_cols:
    if col in cleaned_df.columns:
        median_value = cleaned_df[col].median()
        cleaned_df[col] = cleaned_df[col].fillna(median_value)
        print(f" -> Success: Overwrote missing entries in [{col}] with Median ({median_value}).")
        
# 2. Categorical Structural Imputation Layer (Utilizing Column Modes)
categorical_cols = ['PaymentMethod', 'OrderStatus', 'Carcoupon', 'ReferralSource']
for col in categorical_cols:
    if col in cleaned_df.columns:
        mode_value = cleaned_df[col].mode()[0]
        cleaned_df[col] = cleaned_df[col].fillna(mode_value)
        print(f" -> Success: Overwrote missing entries in [{col}] with Mode ('{mode_value}').")
        
print("\n✔️ System Alert: Imputation validation pipeline completely executed!")
```

## **🧬 Section 3: Multi-Variable Transactional Identity Verification Audit (Step 3)**

### **🔹 Core Objective & Technical Logic**

#### Data synchronization faults across distributed transactional channels dump duplicate logs into backend database environments. This subsystem functions as an enterprise-grade deduplication gate, evaluating structural indices across a compound dimensional spectrum to enforce record row uniqueness.

* **Composite Subset Matrix Keys:** Locks multiple parameters simultaneously (`OrderID`, `Date`, and `CustomerID`) to verify identity.

* **Initial Instance Preservation:** Drops duplicate historical trails while retaining the absolute initial transaction record entry (`keep='first'`).

```python
print("🛑 Step 3: Launching identity verification audit matrix for duplicate isolation...")

# Establish compound key array configurations to isolate record uniqueness parameters
composite_keys = ['OrderID', 'Date', 'CustomerID']
valid_keys = [key for key in composite_keys if key in cleaned_df.columns]

# Capture pre-audit matrix sequence count parameters
initial_count = len(cleaned_df)

# Execute compound deduplication pass, preserving the first chronological record
cleaned_df = cleaned_df.drop_duplicates(subset=valid_keys, keep='first')
final_count = len(cleaned_df)

print(f" -> Verification Audit Success: Discovered and purged {initial_count - final_count} duplicate record sets.")
print(f" -> Stabilized transactional base size: {final_count} active unique rows retained.")
```

## **🌐 Section 4: Global Transformation, Chronological Standardization & Feature Derivation (Step 4)**

### **🔹 Core Objective & Technical Logic**

#### This module handles globalization alignment and mathematical feature engineering. It forces chaotic date layouts into an immutable ISO 8601 framework string, cleans unorganized text inputs into uniform title casing, and engineers a custom continuous variable tracking vector.

* **Chronological Alignment:** Standardizes all timeline events to explicit **ISO 8601 (`YYYY-MM-DD`)** arrays.

* **Calculated Feature Layer:** Derives `TotalPrice` by performing multi-column matrix multiplication, capped to strict two-decimal precision.

```python
print("⚙️ Step 4: Enforcing ISO 8601 Formatting & Text Normalization...")

# 1. Date Format Standardization to ISO 8601 (YYYY-MM-DD)
if 'Date' in cleaned_df.columns:
    cleaned_df['Date'] = pd.to_datetime(cleaned_df['Date'], errors='coerce')
    cleaned_df['Date'] = cleaned_df['Date'].dt.strftime('%Y-%m-%d')
    print(" -> Chronological values converted to uniform 'YYYY-MM-DD' framework.")
    
# 2. Text Spacing & Title Case Normalization
text_targets = ['Product', 'PaymentMethod', 'OrderStatus', 'ReferralSource']
for col in text_targets:
    if col in cleaned_df.columns:
        cleaned_df[col] = cleaned_df[col].astype(str).str.strip().str.title()
print(" -> Text variables synchronized to formal Title-Case layout.")

# 3. Calculated Precision Feature Engineering Layer
if 'UnitPrice' in cleaned_df.columns and 'Quantity' in cleaned_df.columns:
    cleaned_df['TotalPrice'] = (cleaned_df['Quantity'] * cleaned_df['UnitPrice']).round(2)
    print(" -> Derived field 'TotalPrice' computed to 2-decimal accuracy.")
    
print("\n📝 Sample Display: First 3 cleaned dataset records:")
print(cleaned_df.head(3))

print("\n📊 Integrity Audit Checklist (Total Nulls left after cleaning procedures):")
print(cleaned_df.isnull().sum())
```
